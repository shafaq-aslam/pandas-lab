# DataFrame Creation

```python
import numpy as np
import pandas as pd
```

#### Using lists

```python
student_data = [
    [95, 85, 10],
    [100, 98, 15],
    [120, 100, 20],
    [80, 75, 3],
    [150, 100, 25]
]

pd.DataFrame(student_data, columns=['IQ', 'Marks', 'Package'])
```

#### Using dicts

```python
student_dict = {
    'IQ': [100, 90, 85, 50, 95],
    'Marks': [99, 95, 65, 20, 85],
    'Package': [12, 10, 8, 3, 11]
}

students = pd.DataFrame(student_dict)
students
```
