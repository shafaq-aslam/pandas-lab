## Selecting Rows from a DataFrome

<ol>
    <li><strong>iloc:</strong> search using index positions</li>
    <li><strong>loc:</strong> search using index labels</li>
</ol>

```python
student_dict = {
    'Name': ['Shayan', 'Basit', 'Ayan', 'Rohit', 'Zain'],
    'IQ': [100, 90, 85, 50, 95],
    'Marks': [99, 95, 65, 20, 85],
    'Package': [12, 10, 8, 3, 11]
}

students = pd.DataFrame(student_dict)
students
```

```python
# students.set_index('Name') # ---> Name will serve as index
```

```python
students.set_index('Name', inplace = True) # ---> to make changes permanent
students
```

```python
students
```

### iloc

```python
# single row
movies.iloc[5]
```

```python
# multiple rows
movies.iloc[0:5]
```

```python
matches.iloc[0:15:2]
```

```python
# fancy indexing
matches.iloc[[0, 5, 10]]
```

```python
students.iloc[0]
```

### loc

```python
# single row
students.loc['Shayan']
```

```python
# multiple rows
students.loc['Shayan':'Zain']
```

```python
students.loc['Shayan':'Zain':2]
```

```python
# fancy indexing
students.loc[['Shayan', 'Basit', 'Rohit']]
```

## Selecting both from a DataFrame

```python
movies.iloc[0:3, 0:3]
```

```python
matches.loc[5:9, 'ID':'Season']
```
