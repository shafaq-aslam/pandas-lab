# DatetimeIndex Object

<p>A collection of pandas timestamp.</p>

```python
import pandas as pd
```

```python
# from strings
pd.DatetimeIndex(['2025/7/11', '2025/7/12', '2025/7/13'])
```

```python
# using python datetime object
import datetime as dt
pd.DatetimeIndex([dt.datetime(2025,7,11), dt.datetime(2025,7,12)])
```

```python
# using pd.timestamps
pd.DatetimeIndex([pd.Timestamp(2025,7,11), pd.Timestamp(2025,7,12)])
```

```python
# using datetimeindex as series index
date_time = pd.DatetimeIndex([pd.Timestamp(2025,7,11), pd.Timestamp(2025,7,12)])
pd.Series([1, 2], index = date_time)
```
