# Time and Date

```python
import numpy as np
import pandas as pd
```

### Timestamp Object

<p>Time stamps reference particular moments in time (e.g., Oct 24th, 2022 at 7:00pm)</p>

## Creating Timestamp Objects

```python
# Creating a timestamp
pd.Timestamp('2025/11/7')
```

```python
# variations
pd.Timestamp('2025-11-7')
```

```python
pd.Timestamp('2025, 11, 7')
```

```python
# only year
pd.Timestamp('2025')
```

```python
# using text
pd.Timestamp('7th November 2025')
```

```python
# provide time also
pd.Timestamp('7th November 2025 9:55AM')
```

```python
pd.Timestamp('2025/11/7 9:55AM')
```

```python
pd.Timestamp('2025-11-7 9:55AM')
pd.Timestamp('2025, 11, 7 9:55AM')
```
