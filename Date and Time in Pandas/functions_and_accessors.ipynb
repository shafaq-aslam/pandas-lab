# date_range function

```python
import pandas as pd
```

```python
# generate daily dates in a given range
pd.date_range(start = '2025/11/7', end = '2025/11/30' ,freq = 'D')
```

```python
# alternate days in a given range
pd.date_range(start = '2025/11/7', end = '2025/11/30' ,freq = '2D')
```

```python
# B -> business days
pd.date_range(start = '2025/11/7', end = '2025/11/30' ,freq = 'B')
```

```python
# W -> one week per day
pd.date_range(start = '2025/11/7', end = '2025/11/30' ,freq = 'W')
```

```python
pd.date_range(start = '2025/11/7', end = '2025/11/30' ,freq = 'W-SAT')
```

```python
# H -> Hourly data(factor)
pd.date_range(start = '2025/11/7', end = '2025/11/30' ,freq = 'H')
```

```python
pd.date_range(start = '2025/11/7', end = '2025/11/30' ,freq = '6H')
```

```python
# M -> Month end
pd.date_range(start = '2025/11/7', end = '2025/11/30' ,freq = 'M')
```

```python
# MS -> Month start
pd.date_range(start = '2025/11/1', end = '2025/11/30' ,freq = 'MS')
```

```python
# YE -> Year end (A is depricated)
pd.date_range(start = '2025/11/7', end = '2029/12/31' ,freq = 'YE')
```

```python
# YS -> Year Start (A is depricated)
pd.date_range(start = '2025/11/7', end = '2029/12/31' ,freq = 'YS')
```

```python
# using periods(number of results)
pd.date_range(start = '2025/11/7', periods = 25 ,freq = 'D')
```

```python
pd.date_range(start = '2025/11/7', periods = 25 ,freq = 'H')
```

```python
pd.date_range(start = '2025/11/7', periods = 25 ,freq = '6H')
```

# to_datetime function

<p>Converts an existing objects to pandas timestamp/datetimeindex object.</p>

```python
# simple series example
series = pd.Series(['2025/11/7', '2025/11/8', '2025/11/9', '2025/11/10'])
pd.to_datetime(series)
```

```python
series = pd.Series(['2025/11/7', '2025/11/8', '2025/11/9', '2025/11/10'])
pd.to_datetime(series).dt.year
```

```python
series = pd.Series(['2025/11/7', '2025/11/8', '2025/11/9', '2025/11/10'])
pd.to_datetime(series).dt.day
```

```python
series = pd.Series(['2025/11/7', '2025/11/8', '2025/11/9', '2025/11/10'])
pd.to_datetime(series).dt.day_name()
```

```python
# with errors
series = pd.Series(['2025/11/7', '2025/11/8', '2025/11/9', '2025/131/10'])
pd.to_datetime(series)
```

```python
# error handling
series = pd.Series(['2025/11/7', '2025/11/8', '2025/11/9', '2025/131/10'])
pd.to_datetime(series, errors = 'coerce')
```

```python
series = pd.Series(['2025/11/7', '2025/11/8', '2025/11/9', '2025/131/10'])
pd.to_datetime(series, errors = 'coerce').dt.month_name()
```
