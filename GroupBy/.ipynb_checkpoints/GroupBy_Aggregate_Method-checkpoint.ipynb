# GroupBy Aggregate Method

```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('imdb-top-1000.csv')
movies.head(2)
```

## by passing dict

```python
genre = movies.groupby('Genre')
genre.sum(numeric_only = True)
```

```python
genre.agg(
    {
        'Runtime' : 'mean',
        'IMDB_Rating' : 'mean',
        'No_of_Votes' : 'sum',
        'Gross' : 'sum',
        'Metascore' : 'min'
    }
)
```

## by passing list

```python
genre[movies.select_dtypes(include = 'number').columns].agg(['min', 'max', 'mean'])
```

```python
# another way
genre[['Runtime', 'IMDB_Rating', 'No_of_Votes', 'Gross', 'Metascore']].agg(['min', 'max', 'mean', 'sum'])
```

## adding both the syntax

```python
genre.agg(
    {
        'Runtime' : ['mean', 'min'],
        'IMDB_Rating' : ['mean', 'max'],
        'No_of_Votes' : ['sum', 'mean'],
        'Gross' : 'sum',
        'Metascore' : ['min', 'mean']
    }
)
```
