# GroupBy Objects

```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('imdb-top-1000.csv')
movies.head(2)
```

```python
genres = movies.groupby('Genre') # groupby --> apply/work on categorical data
genres
```

```python
# Applying builtin aggregation functions on groupby objects
genres.sum(numeric_only=True)
```

```python
genres.min()
```

```python
genres.mean(numeric_only=True)
```

```python
# Find the top 3 genres by total earning
movies.groupby('Genre')
```

```python
movies.groupby('Genre').sum(numeric_only=True)
```

```python
movies.groupby('Genre').sum()['Gross']
```

```python
movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False)
```

```python
movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3) # final ans
```

```python
# another way --> this is good approach --> fast
movies.groupby('Genre')['Gross']
```

```python
movies.groupby('Genre')['Gross'].sum()
```

```python
movies.groupby('Genre')['Gross'].sum().sort_values(ascending=False).head(3)
```
