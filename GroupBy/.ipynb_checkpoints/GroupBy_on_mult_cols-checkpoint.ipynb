# GroupBy on Multiple Columns

```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('imdb-top-1000.csv')
movies.head(2)
```

```python
movie = movies.groupby(['Director', 'Star1'])
movie
```

```python
# size()
movie.size()
```

```python
# get_group()
movie.get_group(('Aamir Khan', 'Amole Gupte'))
```

```python
# Find the most earning actor-director combo
actor_director_combo = movie['Gross'].sum()
actor_director_combo
```

```python
actor_director_combo.sort_values(ascending = False).head(1)
```

```python
# Find the best(in-terms of metascore(avg)) actor-genre combo
actor_genre_combo = movies.groupby(['Star1', 'Genre'])['Metascore'].mean()
actor_genre_combo
```

```python
# this is multi index series
actor_genre_combo.reset_index() # to convert it DF
```

```python
a_g_combo = actor_genre_combo.reset_index().sort_values('Metascore', ascending = False)
a_g_combo
```

```python
a_g_combo.head(1)
```

```python
# agg on multiple groupby
movie[movies.select_dtypes(include = 'number').columns].agg(['min', 'max', 'mean'])
```

### Excecise

```python
ipl_data = pd.read_csv('deliveries.csv')
ipl_data.head()
```

```python
ipl_data.shape
```

```python
# Find the top 10 batsman in terms of runs
ipl_groupby = ipl_data.groupby('batsman')
ipl_groupby
```

```python
ipl_groupby['batsman_runs'].sum()
```

```python
ipl_groupby['batsman_runs'].sum().sort_values(ascending =False).head(10)
```

```python
# Find the batsman with max no of sixes
sixers = ipl_data[ipl_data['batsman_runs'] == 6]
sixers
```

```python
sixers.groupby('batsman')['batsman'].count().sort_values(ascending = False).head(1)
```

```python
# Find batsman with most number of 4's and 6's in last 5 overs
ipl_over_df = ipl_data[ipl_data['over'] > 15]
ipl_over_df
```

```python
ipl_over_df = ipl_over_df[(ipl_over_df['batsman_runs'] == 4) | (ipl_over_df['batsman_runs'] == 6)]
ipl_over_df
```

```python
ipl_over_df.groupby('batsman')['batsman'].count().sort_values(ascending = False).head(1)
```

```python
ipl_over_df.groupby('batsman')['batsman'].count().sort_values(ascending = False).head(1).index[0]
```
