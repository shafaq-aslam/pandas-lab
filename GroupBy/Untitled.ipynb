# GroupBy Attributes and Methods

```python
# Find total number of groups
print(f"By using len: {len(movies.groupby('Genre'))}") # faster
print(f"By using nunique: {movies['Genre'].nunique()}")
```

```python
# Find items in each group
print(f"{movies.groupby('Genre').size()}")
```

```python
print(f"{movies['Genre'].value_counts()}")
```

```python
# First()/last() item of every group
movies.groupby('Genre').first()
```

```python
movies.groupby('Genre').last()
```

```python
movies.groupby('Genre').nth(6)
```

```python
# get_group
movies.groupby('Genre').get_group('Horror')
```
