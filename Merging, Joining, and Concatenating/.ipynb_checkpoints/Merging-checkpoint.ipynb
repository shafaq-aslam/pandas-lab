# Merging

```python
import numpy as np, pandas as pd
```

```python
courses = pd.read_csv('courses.csv')
students = pd.read_csv('students.csv')
reg_month1 = pd.read_csv('reg-month1.csv')
reg_month2 = pd.read_csv('reg-month2.csv')
matches = pd.read_csv('matches.csv')
deliveries = pd.read_csv('deliveries.csv')
```

```python
registrations = pd.concat([reg_month1, reg_month2], ignore_index = True)
```

### inner join

```python
# you can merge 2 dataframe (students and registration) by using inner join on the basis of student_id, 
# because this is the common column on the both dataframes

students.merge(registrations, how = 'inner', on = 'student_id')
```

### left join

```python
courses.merge(registrations, how = 'left', on = 'course_id')
```

### right join

```python
temp_df = pd.DataFrame({
    'student_id' : [26, 27, 28],
    'name' : ['Shaya', 'Waqas', 'Aslam'],
    'partner' : [28, 26, 17]
})

students = pd.concat([students, temp_df], ignore_index = True)
students
```

```python
students.merge(registrations, how = 'right', on = 'student_id')
```

```python
students.merge(registrations, how = 'left', on = 'student_id')
```

### outer join

```python
students.merge(registrations, how = 'outer', on = 'student_id')
```

```python
# Alternate syntax for merge
pd.merge(students, registrations, how = 'inner', on = 'student_id')
```
