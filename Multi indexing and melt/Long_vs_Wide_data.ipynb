# Long Vs Wide Data

<img src = "wideLong.png" alt = "Wide vs Long table" width = "650">

<p>In pandas, data can be structured in either a long format or a wide format, each suitable for different analytical tasks.</p>

<h1>Wide Format:</h1>
<ol>
    <li><h3>Structure: </h3>Each row represents a unique observation, and different attributes or variables are stored in separate columns.</li>
    <li><h3>Characteristics:</h3>
        <ul>
            <li>More columns, fewer rows (relative to long format).</li>
            <li>Column headers often represent values or categories of a variable.</li>
            <li>Easier for human readability and direct analysis of individual attributes.</li>
        </ul>
    </li>
    <li><h3>Example: </h3>A dataset with each row representing a person, and columns for "Age", "Height", "Weight".</li>
</ol>

<h1>Long Format:</h1>
<ol>
    <li><h3>Structure: </h3>Each row represents a single observation of a specific attribute, with identifier columns to group related observations.</li>
    <li><h3>Characteristics:</h3>
        <ul>
            <li>Fewer columns, more rows (relative to wide format).</li>
            <li>Typically includes an "identifier" column, a "variable" column (naming the attribute), and a "value" column.</li>
            <li>More suitable for statistical analysis, plotting, and database storage.</li>
        </ul>
    </li>
    <li><h3>Example: </h3>A dataset with each row representing a measurement for a person (e.g., one row for "Person A, Age, 30", another for "Person A, Height, 175").</li>
</ol>
