# Vectorized Operations

```python
import numpy as np
import pandas as pd
```

```python
# What are vectorized operations
arr1 = np.array([1,2,3,4,5])
arr1*4 # this is the example of vectorized operations
```

```python
# problem in vectorized operations in vanila python
st = ['cat', 'mat', None, 'rat']
[i.startswith('c') for i in st] 
# this is problem, if your data has none value so python can't handle it
# and another problem is it is slow.
```

```python
# How pandas solves this issue?
st = pd.Series(['cat', 'mat', None, 'rat'])
# str is string accessor
st.str.startswith('c') 
# this is how pandas solve the above issue, and it is relatively fast for the larger datasets.
```

```python
df = pd.read_csv('titanic.csv')
df.sample(5)
```

```python
df['Name']
```

## Common Functions

```python
# lower/upper/capitalize/title
df['Name'].str.lower()
```

```python
df['Name'].str.upper()
```

```python
df['Name'].str.capitalize()
```

```python
df['Name'].str.title()
```

```python
# len
df['Name'].str.len()
```

```python
df['Name'].str.len().max()
```

```python
df['Name'][df['Name'].str.len() == 82].values[0]
```

```python
# strip
"            shayan              ".strip()
df['Name'].str.strip()
```

```python
# split --> get
df['Last_Name'] = df['Name'].str.split(',').str.get(0)
df.sample(5)
```

```python
df[['Title', 'First_Name']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n = 1, expand = True)
df.head(5)
```

```python
df['Title'].value_counts()
```

```python
# replace
df['Title'] = df['Title'].str.replace('Ms.', 'Miss.')
df['Title'] = df['Title'].str.replace('Mlle.', 'Miss.')
df['Title'].value_counts()
```

```python
# filtering (startswith/endswith)
df[df['First_Name'].str.startswith('S')]
```

```python
df[df['First_Name'].str.endswith('E')]
```

```python
# isdigit/isalpha
df[df['First_Name'].str.isdigit()]
```

```python
# contains
# search john --> both case
df[df['First_Name'].str.contains('john', case = False)]
```

```python
# Find lastnames with start and end char vowel
df[df['Last_Name'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')] # regular expression
```

```python
df[df['Last_Name'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```
